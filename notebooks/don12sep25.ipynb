```python
import pandas as pd
import numpy as np
```

```python
fpath = '../data/prep/'
dfile = 'form_data_v2.2.csv'
```

```python
df = pd.read_csv( fpath + dfile, header=0, parse_dates=True )
```

```python
# flat races
df = df[df.Type.isna()]
```

```python
# sandown
df_don = df[df.CourseName=='Doncaster']
```

```python
# 2.00 5f GS/Sft Md 2yo Cl4 5k
df_115 = df_don[((df_don.Distance>14.5*220-55)&(df_don.Distance<14.5*220+55))&((df_don.Going1=='goodtosoft')|(df_don.Going1=='good'))
    &(df_don.RT=='Hc')]
df_115[(df_115.Posn=='1')|(df_115.Posn=='2')][['Sp','Comments']]
```
