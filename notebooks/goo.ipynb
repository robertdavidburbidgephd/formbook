```python
import pandas as pd
import numpy as np
```

```python
fpath = '../data/prep/'
dfile = 'form_data_v2.0.csv'
```

```python
df = pd.read_csv( fpath + dfile, header=0, parse_dates=True )
```

```python
goo = df[(df.CourseName=='Goodwood')&(df.Going1=='goodtosoft')&(df.RT.str.contains('Hc'))&(df.Distance.apply(lambda x: (x>6*220-55) & (x<6*220+55)))]
```

```python
# no. declared from max draw by race, ie FieldSize + RR + withdrawn
DeclFieldSize = goo[['RaceId', 'Draw']].groupby('RaceId').max().squeeze().rename('DeclFieldSize').astype(int)
goo = goo.join(DeclFieldSize, on='RaceId')
```

```python
goo['Posn'] = goo.apply(lambda r: int(r.Posn) if r.Posn.isnumeric() else r.DeclFieldSize, axis=1)
```
